```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import *
```

```python
spark = SparkSession.builder \
    .appName("Week6 Assignment") \
    .master("local[*]") \
    .getOrCreate()
```

```python
df = spark.read.csv(
    r"source_dataset.csv",
    header=True,
    inferSchema=True
)
```

```python
df.show(10)
```

```python
df.printSchema()
```

```python
print("Total Rows :", df.count())
print("Total Columns :", len(df.columns))
print(df.columns)
```

```python
selected_df = df.select(
    "order_id",
    "product_name",
    "category",
    "price",
    "quantity",
    "status"
)

selected_df.show(10)
```

```python
completed_df = selected_df.filter(
    col("status") == "Completed"
)

completed_df.show()
```

```python
electronics_df = completed_df.filter(
    col("category") == "Electronics"
)

electronics_df.show()
```

```python
renamed_df = electronics_df.withColumnRenamed(
    "product_name",
    "Product_Name"
)

renamed_df.show()
```

```python
cast_df = renamed_df.withColumn(
    "price",
    col("price").cast("double")
)

cast_df.printSchema()
```

```python
gst_df = cast_df.withColumn(
    "final_price",
    col("price") * 1.18
)

gst_df.show()
```

```python
total_df = gst_df.withColumn(
    "total_amount",
    col("price") * col("quantity")
)

total_df.show()
```

```python
total_df.select(
    [count(when(col(c).isNull(), c)).alias(c) for c in total_df.columns]
).show()
```

```python
clean_df = total_df.na.fill(0)

clean_df.show()
```

```python
sorted_df = clean_df.orderBy(
    col("price").desc()
)

sorted_df.show()
```

```python
group_df = sorted_df.groupBy("category").agg(
    count("*").alias("Total_Orders"),
    round(avg("price"),2).alias("Average_Price")
)

group_df.show()
```

```python
group_df.cache()

group_df.show()
```

```python
clean_df.write.mode("overwrite") \
.option("header",True) \
.csv("output_csv")
```

```python
clean_df.write.mode("overwrite") \
.parquet("output_parquet")
```

```python
parquet_df = spark.read.parquet(
    "output_parquet"
)

parquet_df.show()
```

```python
print(
    "Partitions :",
    parquet_df.rdd.getNumPartitions()
)
```
